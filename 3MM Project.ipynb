# Three-moment matching for a basket of Black forwards

Approximate $B_T=\sum_{i=1}^n w_iF_i(T)$ by a shifted lognormal with the same first three moments. Part 1 derives the parameters; Part 2 tests the approximation.


## Part 1. Analytical moment matching
### 1.1 Model and notation

For fixed $T>0$, assume under one common probability measure

$$dF_i(t)=\sigma_iF_i(t)\,dW_i(t),\qquad F_i(0)=f_i>0,\qquad dW_i(t)dW_j(t)=\rho_{ij}\,dt.$$

The weights $w_i\in\mathbb R$ and volatilities $\sigma_i\geq0$ are constant. The Brownian correlation matrix $R=(\rho_{ij})$ is symmetric, positive semidefinite, with diagonal one. No normalization of the weights is required.

The Black solution is

$$F_i(T)=f_i e^{-\sigma_i^2T/2+\sigma_i\sqrt T X_i},\qquad X\sim N(0,R).$$

Write $a_i=w_if_i$, $g_{ij}=T\sigma_i\sigma_j\rho_{ij}$, and $m_r=\mathbb E[B_T^r]$.
We use the standard Gaussian MGF:

$$\mathbb E[e^{b^{\mathsf T}X}]=e^{b^{\mathsf T}Rb/2};\qquad
\mathbb E[e^{tZ}]=e^{t^2/2}\quad\text{for }Z\sim N(0,1).$$


### 1.2 The first three basket moments

**First moment.** The Gaussian expectation cancels the drift correction:

$$\mathbb E[F_i(T)]=f_i e^{-g_{ii}/2}e^{g_{ii}/2}=f_i,
\qquad \boxed{m_1=\sum_i a_i.}$$

**Second moment.** Since $\operatorname{Var}(\sqrt T(\sigma_iX_i+\sigma_jX_j))=g_{ii}+g_{jj}+2g_{ij}$,

$$\begin{aligned}
\mathbb E[F_i(T)F_j(T)]
&=f_if_j e^{-(g_{ii}+g_{jj})/2}e^{(g_{ii}+g_{jj}+2g_{ij})/2}\\
&=f_if_j e^{g_{ij}}.
\end{aligned}$$

Expanding $B_T^2$ and taking expectations gives

$$\boxed{m_2=\sum_{i,j}w_iw_j\mathbb E[F_i(T)F_j(T)]
=\sum_{i,j}a_ia_j e^{g_{ij}}.}$$

**Third moment.** The variance of $\sqrt T(\sigma_iX_i+\sigma_jX_j+\sigma_kX_k)$ is
$g_{ii}+g_{jj}+g_{kk}+2(g_{ij}+g_{ik}+g_{jk})$. Thus

$$\begin{aligned}
\mathbb E[F_i(T)F_j(T)F_k(T)]
&=f_if_jf_k e^{-(g_{ii}+g_{jj}+g_{kk})/2}
e^{(g_{ii}+g_{jj}+g_{kk})/2+g_{ij}+g_{ik}+g_{jk}}\\
&=f_if_jf_k e^{g_{ij}+g_{ik}+g_{jk}}.
\end{aligned}$$

Expanding $B_T^3$ gives

$$\boxed{m_3=\sum_{i,j,k}a_ia_ja_k e^{g_{ij}+g_{ik}+g_{jk}}.}$$

All sums are ordered and include repeated indices. In particular,
$\mathbb E[F_i(T)^2]=f_i^2e^{g_{ii}}$ and $\mathbb E[F_i(T)^3]=f_i^3e^{3g_{ii}}$.

For matching, convert to the mean, variance, third central moment, and skewness:

$$\begin{aligned}
m&=m_1,\qquad v=m_2-m^2,\\
c_3&=m_3-3mm_2+2m^3,\qquad \gamma=\frac{c_3}{v^{3/2}}\quad(v>0).
\end{aligned}$$


### 1.3 Moments of the shifted-lognormal approximation

Let

$$\widehat B_T=\widehat C+\widehat F Y,\qquad
Y=e^{-s/2+\sqrt s Z},\qquad s=\widehat\sigma^2T,\quad q=e^s,\quad Z\sim N(0,1).$$

For $r=1,2,3$, the Gaussian MGF gives

$$\mathbb E[Y^r]=e^{-rs/2}e^{r^2s/2}=e^{r(r-1)s/2},
\qquad \mathbb E[Y]=1,\quad\mathbb E[Y^2]=q,\quad\mathbb E[Y^3]=q^3.$$

Expanding the powers of $\widehat C+\widehat F Y$ therefore gives

$$\boxed{\begin{aligned}
\widehat m_1&=\widehat C+\widehat F,\\
\widehat m_2&=\widehat C^2+2\widehat C\widehat F+\widehat F^2q,\\
\widehat m_3&=\widehat C^3+3\widehat C^2\widehat F
+3\widehat C\widehat F^2q+\widehat F^3q^3.
\end{aligned}}$$

The original expression is the special case $\widehat F=1$. Matching its mean and variance fixes
$\widehat C=m-1$ and $q=1+v$, hence forces $c_3=q^3-3q+2=v^2(v+3)$.
The adjustable scale $\widehat F$ is therefore necessary for general three-moment matching.


### 1.4 Match the moments and solve

Since $\widehat B_T-\mathbb E[\widehat B_T]=\widehat F(Y-1)$, matching the first three raw moments is equivalent to

$$\begin{aligned}
m&=\widehat C+\widehat F,\\
v&=\widehat F^2(q-1),\\
c_3&=\widehat F^3(q^3-3q+2)=\widehat F^3(q-1)^2(q+2).
\end{aligned}$$

Assume $v>0$ and $c_3>0$, and seek $\widehat F>0$. Dividing the third equation by the second to power $3/2$ eliminates the scale:

$$\gamma=(q+2)\sqrt{q-1}.$$

Set $r=\sqrt{q-1}>0$. Then $r^3+3r=\gamma$, whose left side increases strictly from zero to infinity. The unique positive root is

$$\boxed{r=2\sinh\!\left[\frac13\operatorname{arsinh}\!\left(\frac\gamma2\right)\right].}$$

This follows from $\sinh(3u)=3\sinh u+4\sinh^3u$. Substituting back gives

$$\boxed{\widehat F=\frac{\sqrt v}{r},\qquad
\widehat C=m-\widehat F,\qquad
\widehat\sigma=\sqrt{\frac{\log(1+r^2)}{T}}.}$$

These parameters match all three analytical moments. They are statistical estimators only when the input moments are estimated from data.


### 1.5 Existence and limitations

* **$v=0$:** the basket is constant; use $\widehat C=m$, $\widehat F=0$, $\widehat\sigma=0$.
* **$v>0$, $c_3=0$:** no finite positive-scale shifted lognormal matches all three moments. A normal distribution arises as the zero-skewness limit.
* **$c_3<0$:** a reflected fit is possible by using $|\gamma|$ in the root formula and $\widehat F=-\sqrt v/r$. The implementation below restricts to positive scale and rejects this case.

For $\widehat F>0$ and $\widehat\sigma>0$, the fit has lower endpoint $\widehat C$, which need not match the basket's support. Three matched moments do not guarantee accurate tails or option values. The fit describes one maturity, not a joint process across maturities.


## Part 2. Implementation and tests
### 2.1 What to test

The simulation draws the exact terminal forwards from the model; there is no time-step error.
We simulate the fitted distribution independently for a visual comparison, but use its
analytical CDF and payoff expectations for quantitative comparisons. This removes
unnecessary simulation noise from the approximation side.

| Check | What it establishes |
|---|---|
| Algebra and exact special cases | Correct implementation, including singular correlations |
| First three empirical moments | Simulator consistency, with Monte Carlo uncertainty |
| Fourth analytical moment | A shape feature not used in fitting |
| CDF distance and a simultaneous confidence bound | Distribution error separated from sampling error |
| Quantiles and fixed-threshold tail probabilities | Lower- and upper-tail discrepancies |
| Calls, puts, and digital payoffs | Error for nonlinear functions of the basket |
| Repeated seeds and stress scenarios | Stability and where the method deteriorates |
| A two-moment lognormal benchmark | Whether matching skewness improves the chosen metric |

Matching three moments is an algebra check, not evidence that the distributions are close.
There is no universally acceptable error threshold: choose tolerances for the intended use.

**Run:** use a Python 3 kernel with NumPy, SciPy, pandas, Matplotlib, and IPython,
then Run All. All example inputs are synthetic; no market data or network access is used.
Defaults: 150,000 main draws, 60,000 per scenario, and 12 independent convergence runs.

### 2.2 Inputs and analytical moments

The loops below follow the double and triple sums directly. Their cost is $O(n^3)$,
which is appropriate for the small example baskets. Input checks prevent invalid
correlation matrices from being silently used.


For stable computation, put $h_{ij}=e^{g_{ij}}-1$. The code uses the equivalent formulas

$$v=\sum_{i,j}a_ia_jh_{ij},\qquad
c_3=\sum_{i,j,k}a_ia_ja_k
(h_{ij}h_{ik}+h_{ij}h_{jk}+h_{ik}h_{jk}+h_{ij}h_{ik}h_{jk}).$$

The second follows by expanding $(1+h_{ij})(1+h_{ik})(1+h_{jk})$ in the raw-moment expression for $c_3$. This avoids subtracting nearly equal raw moments; signed weights can still cause cancellation.


```python
def check_inputs(f0, weights, vol, corr, T):
    # Convert inputs to arrays and reject invalid Black-model parameters.
    f0 = np.asarray(f0, dtype=float)
    weights = np.asarray(weights, dtype=float)
    vol = np.asarray(vol, dtype=float)
    corr = np.asarray(corr, dtype=float)
    if f0.ndim != 1 or len(f0) == 0:
        raise ValueError("f0 must be a nonempty one-dimensional array.")
    n = len(f0)
    if weights.shape != (n,) or vol.shape != (n,) or corr.shape != (n, n):
        raise ValueError("Input dimensions do not agree.")
    if not all(np.isfinite(x).all() for x in (f0, weights, vol, corr)):
        raise ValueError("All inputs must be finite.")
    if not np.isfinite(T) or T <= 0 or np.any(f0 <= 0) or np.any(vol < 0):
        raise ValueError("Require T > 0, positive initial forwards, and nonnegative volatilities.")
    if not np.allclose(corr, corr.T, atol=1e-12, rtol=0):
        raise ValueError("Correlation matrix must be symmetric.")
    if not np.allclose(np.diag(corr), 1, atol=1e-12, rtol=0):
        raise ValueError("Correlation matrix diagonal must be one.")
    if np.linalg.eigvalsh(corr).min() < -1e-12:
        raise ValueError("Correlation matrix must be positive semidefinite.")
    return f0, weights, vol, corr, float(T)


def basket_moments(f0, weights, vol, corr, T):
    # Calculate the raw and central moments derived in Part 1.
    f0, weights, vol, corr, T = check_inputs(f0, weights, vol, corr, T)
    a = weights * f0
    g = T * np.outer(vol, vol) * corr
    h = np.expm1(g)  # exp(g) - 1, accurate even when g is small.
    n = len(a)
    mean = float(a.sum())
    m2 = 0.0
    m3 = 0.0
    variance = 0.0
    central3 = 0.0
    variance_abs_sum = 0.0
    central3_abs_sum = 0.0

    for i in range(n):
        for j in range(n):
            m2 += a[i] * a[j] * np.exp(g[i, j])
            term2 = a[i] * a[j] * h[i, j]
            variance += term2
            variance_abs_sum += abs(term2)
            for k in range(n):
                m3 += a[i] * a[j] * a[k] * np.exp(g[i, j] + g[i, k] + g[j, k])
                hij, hik, hjk = h[i, j], h[i, k], h[j, k]
                term3 = a[i] * a[j] * a[k] * (
                    hij * hik + hij * hjk + hik * hjk + hij * hik * hjk)
                central3 += term3
                central3_abs_sum += abs(term3)

    # Cancellation can be severe for offsetting long and short positions.
    eps = np.finfo(float).eps
    if variance < -100 * eps * variance_abs_sum:
        raise ArithmeticError("Negative variance: numerical cancellation is too large.")
    if variance_abs_sum and abs(variance) <= 100 * eps * variance_abs_sum:
        raise ArithmeticError("Variance is unresolved in double precision; use higher precision.")
    if central3_abs_sum and abs(central3) <= 100 * eps * central3_abs_sum:
        raise ArithmeticError("Third central moment is unresolved; use higher precision.")
    if not np.isfinite([mean, m2, m3, variance, central3]).all():
        raise ArithmeticError("Moments overflowed; reduce the parameters or use higher precision.")

    skewness = central3 / variance**1.5 if variance > 0 else np.nan
    return {"mean": mean, "m2": m2, "m3": m3, "variance": variance,
            "central3": central3, "skewness": skewness}


def fit_three_moments(moments, T):
    # Fit a positive-scale shifted lognormal, with an explicit constant case.
    m, v, c3 = moments["mean"], moments["variance"], moments["central3"]
    if T <= 0 or not np.isfinite([m, v, c3, T]).all() or v < 0:
        raise ValueError("Invalid moments or maturity.")
    if v == 0:
        if c3 != 0:
            raise ValueError("Zero variance requires zero third central moment.")
        return {"C": m, "F_hat": 0.0, "sigma": 0.0, "s": 0.0, "q": 1.0,
                "r": 0.0, "mean": m}
    if c3 <= 0:
        raise ValueError("A nondegenerate positive-scale shifted lognormal requires c3 > 0.")

    gamma = c3 / v**1.5
    if gamma < 1e-7:
        warnings.warn("Very small skewness: the shift and scale are poorly conditioned.")
    r = 2 * np.sinh(np.arcsinh(gamma / 2) / 3)
    s = np.log1p(r * r)
    F_hat = np.sqrt(v) / r
    if not np.isfinite([r, s, F_hat]).all() or r <= 0:
        raise ArithmeticError("The fit is not resolved in double precision.")
    return {"C": m - F_hat, "F_hat": F_hat, "sigma": np.sqrt(s / T), "s": s,
            "q": 1 + r*r, "r": r, "mean": m}


def fitted_raw_moments(fit):
    # Check the three explicit raw-moment formulas independently of simulation.
    C, F_hat, q = fit["C"], fit["F_hat"], fit["q"]
    return np.array([C + F_hat, C*C + 2*C*F_hat + F_hat*F_hat*q,
                     C**3 + 3*C*C*F_hat + 3*C*F_hat*F_hat*q + F_hat**3*q**3])
```

```python
# Synthetic example. Volatility 0.20 means 20% per square root of a year.
f0 = np.array([100.0, 95.0, 105.0])
weights = np.array([0.40, 0.35, 0.25])
vol = np.array([0.20, 0.30, 0.40])
corr = np.array([[1.00, 0.50, 0.25],
                 [0.50, 1.00, 0.40],
                 [0.25, 0.40, 1.00]])
T = 1.0
discount_factor = 1.0  # Undiscounted expected payoffs; replace with D(0,T) for prices.

moments = basket_moments(f0, weights, vol, corr, T)
fit = fit_three_moments(moments, T)
display(pd.Series(moments, name="Analytical basket moments").to_frame())
display(pd.Series({key: fit[key] for key in ["C", "F_hat", "sigma"]},
                  name="Three-moment fit").to_frame())

raw_basket = np.array([moments["mean"], moments["m2"], moments["m3"]])
raw_fit = fitted_raw_moments(fit)
display(pd.DataFrame({"basket": raw_basket, "fit": raw_fit,
                      "relative_difference": (raw_fit - raw_basket) / raw_basket},
                     index=["E[B]", "E[B^2]", "E[B^3]"]))
np.testing.assert_allclose(raw_fit, raw_basket, rtol=2e-12, atol=1e-9)

# Show why the original unit-scale expression fails for these same inputs.
original_c3 = moments["variance"]**2 * (moments["variance"] + 3)
print(f"Basket third central moment: {moments['central3']:,.6f}")
print(f"Original unit-scale model after matching mean/variance: {original_c3:,.6f}")
```

### 2.3 Exact terminal simulation and the analytical fitted distribution

Use $X=LU$, with $LL^{\mathsf T}=R$. An eigenvalue factorization supports positive
semidefinite matrices, including perfect correlation; a Cholesky factor need not exist
in the singular case. Only negative eigenvalues within the input-check tolerance are
clipped to zero.

For $x>\widehat C$, solving $\widehat C+\widehat F e^{-s/2+\sqrt sZ}\leq x$ gives

$$F_{\widehat B}(x)=\Phi\!\left(\frac{\log((x-\widehat C)/\widehat F)+s/2}{\sqrt s}\right),
\qquad Q_{\widehat B}(p)=\widehat C+\widehat F e^{-s/2+\sqrt s\Phi^{-1}(p)}.$$

For $x\leq\widehat C$ the CDF is zero in the nondegenerate case.
In SciPy, use shape $\sqrt s$, location $\widehat C$, and scale
$\widehat F e^{-s/2}$ ([lognorm documentation](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.lognorm.html)).


```python
def simulate_basket(f0, weights, vol, corr, T, n_draws, rng):
    # Independent draws of the exact terminal basket, without Euler time steps.
    f0, weights, vol, corr, T = check_inputs(f0, weights, vol, corr, T)
    eigenvalues, eigenvectors = np.linalg.eigh(corr)
    L = eigenvectors @ np.diag(np.sqrt(np.maximum(eigenvalues, 0)))
    independent_normals = rng.standard_normal((n_draws, len(f0)))
    correlated_normals = independent_normals @ L.T
    terminal_forwards = f0 * np.exp(-0.5 * vol**2 * T + vol * np.sqrt(T) * correlated_normals)
    return terminal_forwards @ weights


def simulate_fit(fit, n_draws, rng):
    z = rng.standard_normal(n_draws)
    # mean + F_hat*(exp(x)-1) avoids cancelling a large negative C and large F_hat.
    return fit["mean"] + fit["F_hat"] * np.expm1(-0.5 * fit["s"] + np.sqrt(fit["s"]) * z)


def fitted_distribution(fit):
    if fit["F_hat"] == 0:
        raise ValueError("The constant case has no continuous lognormal distribution.")
    return stats.lognorm(s=np.sqrt(fit["s"]), loc=fit["C"],
                         scale=fit["F_hat"] * np.exp(-0.5 * fit["s"]))


def fit_two_moments(moments, T):
    # An ordinary lognormal benchmark, defined only for a positive mean.
    m, v = moments["mean"], moments["variance"]
    if m <= 0 or v <= 0:
        raise ValueError("Two-moment lognormal requires positive mean and variance.")
    s = np.log1p(v / m**2)
    return {"C": 0.0, "F_hat": m, "sigma": np.sqrt(s / T), "s": s,
            "q": np.exp(s), "r": np.sqrt(v) / m, "mean": m}


# Separate random streams make the two simulation samples independent.
basket_seed, fit_seed = np.random.SeedSequence(SEED).spawn(2)
basket_sample = simulate_basket(f0, weights, vol, corr, T, N_MAIN,
                                np.random.default_rng(basket_seed))
fit_sample = simulate_fit(fit, N_MAIN, np.random.default_rng(fit_seed))
fit_dist = fitted_distribution(fit)
fit_2mm = fit_two_moments(moments, T)
dist_2mm = fitted_distribution(fit_2mm)
```

### 2.4 Exact cases and failure cases

With one forward and positive weight, the basket is exactly lognormal: $C=0$,
$\widehat F=wf_0$, and $\widehat\sigma=\sigma$.
If all volatilities are equal and all correlations are one, the same conclusion holds
with $\widehat F=\sum_iw_if_i>0$. A deterministic positive holding plus one lognormal holding
is exactly shifted lognormal.

These checks test known answers. They do not assert that a Monte Carlo sample must
equal its population moments or pass a random significance test.


```python
# 1. A single forward.
one = basket_moments([100], [1], [0.30], [[1]], 2.0)
one_fit = fit_three_moments(one, 2.0)
np.testing.assert_allclose([one_fit["C"], one_fit["F_hat"], one_fit["sigma"]],
                           [0, 100, 0.30], atol=1e-10)

# 2. Perfect correlation and equal volatilities: singular R is valid.
perfect = basket_moments(f0, weights, np.full(3, 0.25), np.ones((3, 3)), T)
perfect_fit = fit_three_moments(perfect, T)
np.testing.assert_allclose([perfect_fit["C"], perfect_fit["F_hat"], perfect_fit["sigma"]],
                           [0, weights @ f0, 0.25], atol=1e-10)

# 3. A constant holding plus one stochastic holding.
shifted = basket_moments([30, 70], [1, 1], [0, 0.25], np.eye(2), T)
shifted_fit = fit_three_moments(shifted, T)
np.testing.assert_allclose([shifted_fit["C"], shifted_fit["F_hat"], shifted_fit["sigma"]],
                           [30, 70, 0.25], atol=1e-10)

# 4. All volatilities zero: the simulator and fit are constants.
constant = basket_moments(f0, weights, np.zeros(3), corr, T)
constant_fit = fit_three_moments(constant, T)
np.testing.assert_allclose(simulate_fit(constant_fit, 20, np.random.default_rng(1)), weights @ f0)

# 5. Negative skewness must not silently produce a positive-scale fit.
negative = basket_moments([100], [-1], [0.30], [[1]], T)
try:
    fit_three_moments(negative, T)
except ValueError as error:
    print("Expected rejection:", error)
else:
    raise AssertionError("Negative skewness was not rejected.")

# 6. A non-PSD correlation matrix is invalid.
try:
    check_inputs([100, 100], [0.5, 0.5], [0.2, 0.2], [[1, 1.2], [1.2, 1]], T)
except ValueError as error:
    print("Expected rejection:", error)
else:
    raise AssertionError("Invalid correlation was not rejected.")

print("All deterministic checks passed.")
```

### 2.5 Simulated moments and a fourth-moment check

For $g(B)=B^r$, estimate its mean by $N^{-1}\sum_jg(B_j)$ and its Monte Carlo
standard error by $s_g/\sqrt N$. The reported $z$-score is the difference from the
analytical value divided by that standard error. This is a simulation diagnostic,
not an exact finite-sample test; high moments can converge slowly.

For the fourth moment, multiplying four forward exponentials leaves one covariance
term for each of the six pairs, exactly as in Section 1.2:

$$m_4=\sum_{i_1,i_2,i_3,i_4}\left(\prod_{u=1}^4a_{i_u}\right)
\exp\!\left(\sum_{1\leq u<v\leq4}g_{i_ui_v}\right).$$

For the fit, write $C=\widehat C$, expand $(C+\widehat FY)^4$, and use $\mathbb E[Y^4]=e^{6s}=q^6$:

$$\widehat m_4=C^4+4C^3\widehat F+6C^2\widehat F^2q+4C\widehat F^3q^3+\widehat F^4q^6.$$

The standardized fourth central moment is
$(m_4-4mm_3+6m^2m_2-3m^4)/v^2$. Unlike the first three moments,
it is not constrained by the fitting equations. For the fit, expansion of $(Y-1)^4$ gives

$$\frac{\widehat c_4}{v^2}=\frac{q^6-4q^3+6q-3}{(q-1)^2}
=q^4+2q^3+3q^2-3.$$

The basket calculation below costs $O(n^4)$
and is intended for these small baskets.


```python
moment_rows = []
for sample_name, sample in [("basket", basket_sample), ("fit", fit_sample)]:
    for order, target in enumerate(raw_basket, start=1):
        values = sample**order
        estimate = values.mean()
        se = values.std(ddof=1) / np.sqrt(len(values))
        moment_rows.append({"sample": sample_name, "order": order,
                            "analytical": target, "MC_estimate": estimate,
                            "MC_SE": se, "difference_in_SE": (estimate - target) / se})
display(pd.DataFrame(moment_rows))


def basket_fourth_moment(f0, weights, vol, corr, T):
    f0, weights, vol, corr, T = check_inputs(f0, weights, vol, corr, T)
    a = f0 * weights
    g = T * np.outer(vol, vol) * corr
    total = 0.0
    for indices in itertools.product(range(len(a)), repeat=4):
        coefficient = np.prod(a[list(indices)])
        exponent = 0.0
        for u in range(4):
            for v in range(u + 1, 4):
                exponent += g[indices[u], indices[v]]
        total += coefficient * np.exp(exponent)
    return total


m4 = basket_fourth_moment(f0, weights, vol, corr, T)
C, F_hat, q = fit["C"], fit["F_hat"], fit["q"]
m4_fit = C**4 + 4*C**3*F_hat + 6*C*C*F_hat*F_hat*q + 4*C*F_hat**3*q**3 + F_hat**4*q**6
m, v = moments["mean"], moments["variance"]
kurtosis_basket = (m4 - 4*m*moments["m3"] + 6*m*m*moments["m2"] - 3*m**4) / v**2
kurtosis_fit = q**4 + 2*q**3 + 3*q*q - 3  # Ordinary kurtosis, not excess kurtosis.
display(pd.DataFrame({"basket": [m4, kurtosis_basket], "fit": [m4_fit, kurtosis_fit]},
                     index=["fourth raw moment", "fourth central moment / variance^2"]))
```

### 2.6 Distribution plots

Histograms compare both independent simulations with the fitted density.
The empirical CDF and Q-Q plot show differences that bin choices can conceal.
The log-scale survival plot makes upper-tail differences visible.
The histogram displays only the central 99.8% region; it is not a full-support comparison.


```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))
lo = min(np.quantile(basket_sample, 0.001), fit_dist.ppf(0.001))
hi = max(np.quantile(basket_sample, 0.999), fit_dist.ppf(0.999))
bins = np.linspace(lo, hi, 75)
x_grid = np.linspace(lo, hi, 700)

# Weights make the bar areas equal probability over bin width, without renormalising the crop.
bin_width = bins[1] - bins[0]
axes[0, 0].hist(basket_sample, bins=bins, weights=np.full(N_MAIN, 1 / (N_MAIN * bin_width)),
                alpha=0.35, label="Basket simulation")
axes[0, 0].hist(fit_sample, bins=bins, weights=np.full(N_MAIN, 1 / (N_MAIN * bin_width)),
                histtype="step", linewidth=1.3, label="Fit simulation")
axes[0, 0].plot(x_grid, fit_dist.pdf(x_grid), label="Analytical fit density")
axes[0, 0].set(title="Density: central 99.8% region", xlabel="Terminal basket", ylabel="Density")

ordered = np.sort(basket_sample)
empirical_cdf = np.searchsorted(ordered, x_grid, side="right") / N_MAIN
axes[0, 1].plot(x_grid, empirical_cdf, label="Basket empirical CDF")
axes[0, 1].plot(x_grid, fit_dist.cdf(x_grid), "--", label="3MM CDF")
axes[0, 1].plot(x_grid, dist_2mm.cdf(x_grid), ":", label="2MM CDF")
axes[0, 1].set(title="Cumulative probabilities", xlabel="Terminal basket", ylabel="CDF")

qq_p = np.linspace(0.001, 0.999, 300)
q_fit = fit_dist.ppf(qq_p)
q_basket = np.quantile(basket_sample, qq_p)
axes[1, 0].plot(q_fit, q_basket, label="Quantile pairs")
axes[1, 0].plot([lo, hi], [lo, hi], "k--", label="Exact agreement")
axes[1, 0].set(title="Q-Q plot", xlabel="Fitted quantile", ylabel="Basket empirical quantile")

tail_grid = np.linspace(moments["mean"], hi, 400)
empirical_survival = 1 - np.searchsorted(ordered, tail_grid, side="right") / N_MAIN
axes[1, 1].semilogy(tail_grid, np.maximum(empirical_survival, 1 / N_MAIN), label="Basket")
axes[1, 1].semilogy(tail_grid, fit_dist.sf(tail_grid), "--", label="3MM")
axes[1, 1].semilogy(tail_grid, dist_2mm.sf(tail_grid), ":", label="2MM")
axes[1, 1].set(title="Upper-tail survival probabilities", xlabel="Threshold", ylabel="P(B > threshold)")
for ax in axes.flat:
    ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
```

### 2.7 CDF distance with a finite-sample uncertainty bound

Let $F_N$ be the empirical basket CDF and $G$ a fixed fitted CDF. Compute
$D_N=\sup_x|F_N(x)-G(x)|$ exactly at both sides of every sample jump.
The target is $D=\sup_x|F_B(x)-G(x)|$.

For independent draws, the Dvoretzky-Kiefer-Wolfowitz bound gives
$\Pr(\sup_x|F_N(x)-F_B(x)|>\varepsilon)\leq2e^{-2N\varepsilon^2}$.
Consequently, with probability at least $1-\alpha$,

$$\max(0,D_N-\varepsilon_N)\leq D\leq\min(1,D_N+\varepsilon_N),\qquad
\varepsilon_N=\sqrt{\frac{\log(2/\alpha)}{2N}}.$$

The interval follows directly from the triangle inequality. It quantifies uncertainty
about an effect size, rather than merely rejecting exact equality.
The same empirical-CDF event covers both fixed benchmark distributions simultaneously.
The probability bound is a statistical tool used in Part 2, not a step in the moment derivation
([Wei and Dudley](https://arxiv.org/abs/1107.5356)).

An ordinary one-sample KS p-value is also reported. Here the distribution is fixed by
analytical model inputs, independently of the sample, so the usual null calibration applies
for a continuous basket distribution. Refitting on the tested data would invalidate that
calibration; a parametric bootstrap would then need to repeat the fitting step
([NIST](https://www.itl.nist.gov/div898/handbook/eda/section3/eda35g.htm)).
A small p-value can correspond to an economically small error when $N$ is large.


```python
def cdf_diagnostics(sample, distribution, alpha=0.05):
    ordered = np.sort(sample)
    n = len(ordered)
    fitted_cdf = distribution.cdf(ordered)
    right_jump = np.arange(1, n + 1) / n - fitted_cdf
    left_jump = fitted_cdf - np.arange(n) / n
    distance = max(right_jump.max(), left_jump.max())
    epsilon = np.sqrt(np.log(2 / alpha) / (2 * n))
    return {"D_empirical": distance, "DKW_epsilon": epsilon,
            "D_lower": max(0.0, distance - epsilon),
            "D_upper": min(1.0, distance + epsilon)}


cdf_rows = []
for label, distribution in [("3MM shifted lognormal", fit_dist), ("2MM lognormal", dist_2mm)]:
    row = cdf_diagnostics(basket_sample, distribution, ALPHA)
    row["model"] = label
    row["KS_pvalue"] = stats.kstest(basket_sample, distribution.cdf, method="asymp").pvalue
    cdf_rows.append(row)
cdf_table = pd.DataFrame(cdf_rows).set_index("model")
display(cdf_table)
print("A zero lower bound does not establish exact agreement.")
```

### 2.8 Quantiles and tail probabilities

Invert the simultaneous empirical-CDF band to bound the true basket quantile:

$$Q_N(p-\varepsilon_N)\leq Q_B(p)\leq Q_N(p+\varepsilon_N).$$

Here $Q_N$ is the generalized inverse of the empirical CDF. Use $-\infty$ or
$+\infty$ when the relevant probability falls outside $(0,1)$.
These conservative bounds are simultaneous over $p$ and can be uninformative in very
extreme tails. The quantile-error interval is
$[Q_{\widehat B}(p)-Q_{\rm upper},\ Q_{\widehat B}(p)-Q_{\rm lower}]$.

For sharper tail checks, choose thresholds from the fitted distribution **before** inspecting
the simulation. Count basket observations below its 1% and 5% quantiles and above its
95% and 99% quantiles. Exact binomial intervals then estimate the actual basket tail
probabilities. Bonferroni adjustment gives at least 95% joint coverage for these four
intervals ([SciPy exact binomial intervals](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats._result_classes.BinomTestResult.proportion_ci.html)).
Zero observed tail events does not establish zero true probability.


```python
def empirical_inverse(ordered, probability):
    if probability <= 0:
        return -np.inf
    if probability >= 1:
        return np.inf
    return ordered[int(np.ceil(len(ordered) * probability)) - 1]


epsilon = np.sqrt(np.log(2 / ALPHA) / (2 * N_MAIN))
quantile_rows = []
basket_sd = np.sqrt(moments["variance"])
for p in [0.001, 0.01, 0.05, 0.50, 0.95, 0.99, 0.999]:
    empirical_q = empirical_inverse(ordered, p)
    fitted_q = fit_dist.ppf(p)
    lower_q = empirical_inverse(ordered, p - epsilon)
    upper_q = empirical_inverse(ordered, p + epsilon)
    quantile_rows.append({"p": p, "basket_MC_quantile": empirical_q, "fit_quantile": fitted_q,
                          "error_in_basket_SD": (fitted_q - empirical_q) / basket_sd,
                          "basket_q_lower": lower_q, "basket_q_upper": upper_q,
                          "fit_minus_true_lower": fitted_q - upper_q,
                          "fit_minus_true_upper": fitted_q - lower_q})
quantile_table = pd.DataFrame(quantile_rows)
display(quantile_table)

tail_rows = []
tail_probabilities = [0.01, 0.05, 0.95, 0.99]
for p in tail_probabilities:
    threshold = fit_dist.ppf(p)
    event = basket_sample <= threshold if p < 0.5 else basket_sample > threshold
    model_probability = p if p < 0.5 else 1 - p
    interval = stats.binomtest(int(event.sum()), N_MAIN).proportion_ci(
        confidence_level=1 - ALPHA / len(tail_probabilities), method="exact")
    tail_rows.append({"event": f"B {'<=' if p < 0.5 else '>'} fitted Q({p})",
                      "threshold": threshold, "events": int(event.sum()),
                      "basket_probability": event.mean(), "fit_probability": model_probability,
                      "joint_CI_lower": interval.low, "joint_CI_upper": interval.high,
                      "fit_outside_CI": not (interval.low <= model_probability <= interval.high)})
tail_table = pd.DataFrame(tail_rows)
display(tail_table)

# The fitted lower endpoint rules out these basket values entirely.
below_support = int(np.sum(basket_sample <= fit["C"]))
support_ci = stats.binomtest(below_support, N_MAIN).proportion_ci(1 - ALPHA, method="exact")
print(f"Basket observations at/below fitted endpoint C={fit['C']:.6g}: {below_support:,}")
print(f"95% interval for this excluded basket probability: [{support_ci.low:.6g}, {support_ci.high:.6g}]")
if np.all(weights >= 0):
    print(f"Fit probability of a negative basket: {fit_dist.cdf(0):.6g}; true probability: 0.")
```

### 2.9 Calls and puts across strikes

These payoffs detect discrepancies in parts of the distribution that moments may miss.
Let $k=K-C$. For $k>0$, the fitted call payoff is positive when
$Z>z_*=[\log(k/\widehat F)+s/2]/\sqrt s$. Completing the square once more,

$$\mathbb E[e^{-s/2+\sqrt sZ}\mathbf1_{Z>z_*}]
=\int_{z_*}^{\infty}\frac{e^{-(z-\sqrt s)^2/2}}{\sqrt{2\pi}}dz
=\Phi(\sqrt s-z_*).$$

Writing $d_2=-z_*$ and $d_1=d_2+\sqrt s$ gives the expected payoffs

$$\mathbb E[(\widehat B_T-K)^+]=\widehat F\Phi(d_1)-k\Phi(d_2),$$
$$\mathbb E[(K-\widehat B_T)^+]=k\Phi(-d_2)-\widehat F\Phi(-d_1).$$

If $K\leq C$, the call expectation is $m-K$ and the put expectation is zero.
Multiply by $D(0,T)$ to obtain prices. The formula applies to the fit, while the true
basket benchmark is Monte Carlo with standard error $s_{\rm payoff}/\sqrt N$.

The table uses fixed strikes $K=m+z\sqrt v$, with $z=-2,-1,0,1,2,3$.
Errors are fit minus basket. Approximate 95% intervals use a normal approximation
and are pointwise, not simultaneous across strikes. Relative errors are omitted
when the Monte Carlo payoff is less than ten standard errors from zero.
The digitals $\mathbf1_{B\leq K}$ or $\mathbf1_{B>K}$ were tested in Section 2.8.


```python
def fitted_payoff(fit, strike, kind="call"):
    # Undiscounted expected call or put payoff under the fitted distribution.
    if kind not in ("call", "put"):
        raise ValueError("kind must be 'call' or 'put'.")
    if fit["F_hat"] == 0:
        return max(fit["mean"] - strike, 0) if kind == "call" else max(strike - fit["mean"], 0)
    k = strike - fit["C"]
    if k <= 0:
        return fit["mean"] - strike if kind == "call" else 0.0
    sqrt_s = np.sqrt(fit["s"])
    d2 = (np.log(fit["F_hat"] / k) - 0.5 * fit["s"]) / sqrt_s
    d1 = d2 + sqrt_s
    if kind == "call":
        return fit["F_hat"] * stats.norm.cdf(d1) - k * stats.norm.cdf(d2)
    return k * stats.norm.cdf(-d2) - fit["F_hat"] * stats.norm.cdf(-d1)


payoff_rows = []
for z in [-2, -1, 0, 1, 2, 3]:
    strike = moments["mean"] + z * basket_sd
    for kind in ["call", "put"]:
        payoff = np.maximum(basket_sample - strike, 0) if kind == "call" else np.maximum(strike - basket_sample, 0)
        mc_price = discount_factor * payoff.mean()
        se = discount_factor * payoff.std(ddof=1) / np.sqrt(N_MAIN)
        approx_price = discount_factor * fitted_payoff(fit, strike, kind)
        benchmark_price = discount_factor * fitted_payoff(fit_2mm, strike, kind)
        error = approx_price - mc_price
        payoff_rows.append({"kind": kind, "strike": strike, "basket_MC": mc_price,
                            "MC_SE": se, "fit_3MM": approx_price, "fit_2MM": benchmark_price,
                            "error_3MM": error, "error_lower": error - 1.96 * se,
                            "error_upper": error + 1.96 * se,
                            "relative_error": error / mc_price if mc_price > 10 * se and mc_price > 0 else np.nan})
payoff_table = pd.DataFrame(payoff_rows)
display(payoff_table)

# Put-call parity is an algebraic check of the payoff formulas.
for strike in [fit["C"] - 1, moments["mean"], moments["mean"] + 3 * basket_sd]:
    np.testing.assert_allclose(fitted_payoff(fit, strike, "call") - fitted_payoff(fit, strike, "put"),
                               moments["mean"] - strike, rtol=1e-12, atol=1e-10)

fig, ax = plt.subplots(figsize=(10, 4.5))
for kind, marker in [("call", "o"), ("put", "s")]:
    selected = payoff_table[payoff_table["kind"] == kind]
    ax.errorbar(selected["strike"], selected["error_3MM"], yerr=1.96 * selected["MC_SE"],
                 fmt=marker + "-", capsize=3, label=kind + ": error with pointwise 95% MC interval")
ax.axhline(0, color="black", linewidth=1)
ax.set(xlabel="Strike", ylabel="3MM minus basket price", title="Payoff errors across strikes")
ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
```

### 2.10 Stress scenarios

Hold the base inputs fixed except for the named change. Exact cases calibrate the simulation
noise; low or negative correlation, heterogeneous volatilities, and longer maturities
test less lognormal basket shapes. Signed weights test the fitted support limitation.

Each row uses fresh random draws. CDF intervals use $\alpha/J$ for $J$ scenarios,
so their joint coverage is at least 95%. The quantile and fourth-moment errors are
effect sizes, not formal significance tests. For calls, $K=m$ within each scenario.


```python
def equal_correlation(n, rho):
    return (1 - rho) * np.eye(n) + rho * np.ones((n, n))


scenarios = [
    ("Base", f0, weights, vol, corr, T),
    ("One forward: exact", [100], [1], [0.30], [[1]], T),
    ("Perfect common factor: exact", f0, weights, np.full(3, 0.25), np.ones((3, 3)), T),
    ("Independent drivers", f0, weights, vol, np.eye(3), T),
    ("Negative correlation", f0, weights, vol, equal_correlation(3, -0.40), T),
    ("High correlation", f0, weights, vol, equal_correlation(3, 0.90), T),
    ("Heterogeneous volatilities", f0, weights, [0.10, 0.30, 0.80], corr, T),
    ("Long maturity: 5 years", f0, weights, vol, corr, 5.0),
    ("Short maturity: 0.05 years", f0, weights, vol, corr, 0.05),
    ("Concentrated weights", f0, [0.90, 0.07, 0.03], vol, corr, T),
    ("Signed spread", [100, 100], [1, -0.8], [0.40, 0.20], equal_correlation(2, 0.30), T),
]

scenario_rows = []
scenario_seeds = np.random.SeedSequence(SEED + 1).spawn(len(scenarios))
for inputs, seed in zip(scenarios, scenario_seeds):
    name, sf0, sw, sv, sr, st = inputs
    sm = basket_moments(sf0, sw, sv, sr, st)
    try:
        sf = fit_three_moments(sm, st)
    except (ValueError, ArithmeticError) as error:
        scenario_rows.append({"scenario": name, "status": str(error)})
        continue
    sample = simulate_basket(sf0, sw, sv, sr, st, N_SCENARIO, np.random.default_rng(seed))
    distribution = fitted_distribution(sf)
    diagnostics = cdf_diagnostics(sample, distribution, ALPHA / len(scenarios))
    sd = np.sqrt(sm["variance"])
    sample_call = np.maximum(sample - sm["mean"], 0)
    call_error = fitted_payoff(sf, sm["mean"]) - sample_call.mean()
    fourth = basket_fourth_moment(sf0, sw, sv, sr, st)
    qq = sf["q"]
    fitted_kurtosis = qq**4 + 2*qq**3 + 3*qq**2 - 3
    true_kurtosis = (fourth - 4*sm["mean"]*sm["m3"] + 6*sm["mean"]**2*sm["m2"] - 3*sm["mean"]**4) / sm["variance"]**2
    benchmark_D = np.nan
    if sm["mean"] > 0:
        benchmark_D = cdf_diagnostics(sample, fitted_distribution(fit_two_moments(sm, st)))["D_empirical"]
    scenario_rows.append({"scenario": name, "status": "fitted", "skewness": sm["skewness"],
                          "D_3MM": diagnostics["D_empirical"], "D_lower_joint": diagnostics["D_lower"],
                          "D_upper_joint": diagnostics["D_upper"], "D_2MM": benchmark_D,
                          "Q99_error_in_SD": (distribution.ppf(0.99) - np.quantile(sample, 0.99)) / sd,
                          "kurtosis_error": fitted_kurtosis - true_kurtosis,
                          "ATM_call_error_in_SD": call_error / sd,
                          "ATM_call_MC_SE_in_SD": sample_call.std(ddof=1) / np.sqrt(N_SCENARIO) / sd,
                          "basket_below_C": np.mean(sample <= sf["C"])})
scenario_table = pd.DataFrame(scenario_rows).set_index("scenario")
display(scenario_table)
```

### 2.11 Repeat the experiment as the sample size grows

Changing the seed should change Monte Carlo noise, not the population approximation.
For each sample size, repeat the base experiment with independent streams.
Report the mean observed CDF distance, its across-run standard deviation, and
the mean ATM call error. A nonzero error that stabilizes as uncertainty shrinks
indicates approximation bias.

Each CDF distance has upward sampling bias because it is a supremum; its across-run
standard deviation is not a confidence interval for the population distance.
The last two columns give an approximate 95% interval for the expected ATM call error
using a Student-$t$ interval across independent runs. The lognormal model has finite
payoff variance, but the finite-sample interval is still approximate.


```python
sample_sizes = [2_000, 10_000, 50_000]
repeat_seeds = np.random.SeedSequence(SEED + 2).spawn(len(sample_sizes) * N_REPEATS)
convergence_rows = []
seed_number = 0
for n_draws in sample_sizes:
    distances = []
    call_errors = []
    for repeat in range(N_REPEATS):
        sample = simulate_basket(f0, weights, vol, corr, T, n_draws,
                                  np.random.default_rng(repeat_seeds[seed_number]))
        seed_number += 1
        distances.append(cdf_diagnostics(sample, fit_dist)["D_empirical"])
        call_errors.append(fitted_payoff(fit, moments["mean"]) - np.maximum(sample - moments["mean"], 0).mean())
    error_mean = np.mean(call_errors)
    error_se = np.std(call_errors, ddof=1) / np.sqrt(N_REPEATS)
    critical = stats.t.ppf(1 - ALPHA / 2, df=N_REPEATS - 1)
    convergence_rows.append({"N_per_run": n_draws, "repeats": N_REPEATS,
                             "D_mean": np.mean(distances), "D_between_run_SD": np.std(distances, ddof=1),
                             "DKW_epsilon_per_run": np.sqrt(np.log(2 / ALPHA) / (2 * n_draws)),
                             "ATM_call_error_mean": error_mean,
                             "ATM_error_lower": error_mean - critical * error_se,
                             "ATM_error_upper": error_mean + critical * error_se})
convergence_table = pd.DataFrame(convergence_rows)
display(convergence_table)
```

### 2.13 Reading the results and extending the experiment

Assess accuracy in this order: known-answer checks, CDF error bounds, tail probabilities,
payoff errors relative to Monte Carlo uncertainty, and stress-case stability.
Do not infer quality from overlapping histograms or matched moments alone.
The 3MM fit is not guaranteed to beat the 2MM fit for every strike or distance measure.

Useful next experiments:

* **Tail averages:** compare $\mathbb E[B_T\mid B_T\leq Q_B(p)]$ and
  $\mathbb E[B_T\mid B_T\geq Q_B(p)]$ with fitted values. Bootstrap the whole
  quantile-and-tail-mean calculation, not just the observations beyond a fixed cutoff.
  These are lower/upper basket tail means; a loss expected shortfall needs a separately defined loss variable.
* **Targeted parameter grids:** vary maturity, volatility dispersion, concentration, and
  correlation one at a time. Record maximum CDF, quantile, and payoff errors against
  prespecified tolerances. The small scenario set here is illustrative, not a universal guarantee.
* **Rare-event precision:** increase independent draws or use importance sampling with
  likelihood weights. Randomized quasi-Monte Carlo can improve efficiency, but ordinary
  i.i.d. binomial and DKW intervals cannot be reused for dependent quasi-random samples.
* **Input uncertainty and sensitivities:** perturb forwards, volatilities, and valid correlation
  matrices, refit each time, and compare price changes against common-random-number
  basket simulations. The present intervals quantify simulation error, not market-calibration error.
* **Alternative families:** compare the normal limit for near-zero skewness, a reflected
  fit for negative skewness, and a richer family when the fourth moment or support is inadequate.

For background on reducing three-moment basket matching to a skewness-based cubic, see
[Hu, Sayit and Viens, *Pricing basket options with the first three moments of the basket*](https://arxiv.org/abs/2302.08041).
Part 1 contains the full derivation used here and does not require that paper's results.
