# Part 1. Three-moment matching

We approximate the basket

$$B_T=\sum_{i=1}^n w_iF_i(T)$$

by

$$\widehat B_T=\widehat C+\widehat F
\exp\!\left(-\frac12\widehat\sigma^2T+\widehat\sigma\sqrt T Z\right),
\qquad Z\sim N(0,1),$$

matching its first three moments. The scale $\widehat F$ supplies the third adjustable parameter.


## 1. Forward dynamics

For $T>0$, constant weights and volatilities, and positive initial forwards, assume under one common probability measure

$$dF_i(t)=\sigma_iF_i(t)\,dW_i(t),\qquad
dW_i(t)dW_j(t)=\rho_{ij}\,dt.$$

The Brownian correlation matrix must be symmetric and positive semidefinite, with diagonal one. The Black solution is

$$F_i(T)=F_i(0)\exp\!\left(-\frac12\sigma_i^2T+\sigma_iW_i(T)\right).$$

Below, Gaussian exponential expectations follow directly from the normal MGF.


## 2. First basket moment

Since $W_i(T)\sim N(0,T)$,

$$\mathbb E[F_i(T)]
=F_i(0)e^{-\sigma_i^2T/2}\mathbb E[e^{\sigma_iW_i(T)}]
=F_i(0)e^{-\sigma_i^2T/2}e^{\sigma_i^2T/2}=F_i(0).$$

Therefore

$$\boxed{\mathbb E[B_T]=\sum_{i=1}^n w_iF_i(0).}$$


## 3. Second basket moment

The Gaussian MGF gives

$$\mathbb E[e^{\sigma_iW_i(T)+\sigma_jW_j(T)}]
=\exp\!\left[\frac T2\left(\sigma_i^2+\sigma_j^2+2\rho_{ij}\sigma_i\sigma_j\right)\right].$$

Multiplying by the drift corrections in the two forwards cancels the individual variance terms:

$$\mathbb E[F_i(T)F_j(T)]
=F_i(0)F_j(0)e^{T\rho_{ij}\sigma_i\sigma_j}.$$

Expanding the basket square gives

$$\boxed{\mathbb E[B_T^2]
=\sum_{i=1}^n\sum_{j=1}^n
w_iw_jF_i(0)F_j(0)e^{T\rho_{ij}\sigma_i\sigma_j}.}$$

Hence

$$\operatorname{Var}(B_T)
=\sum_{i=1}^n\sum_{j=1}^n w_iw_jF_i(0)F_j(0)
\left(e^{T\rho_{ij}\sigma_i\sigma_j}-1\right).$$


## 4. Third basket moment

For three forwards, the Gaussian MGF gives

$$\begin{aligned}
&\mathbb E[e^{\sigma_iW_i(T)+\sigma_jW_j(T)+\sigma_kW_k(T)}]\\
&\quad=\exp\!\left[\frac T2(\sigma_i^2+\sigma_j^2+\sigma_k^2)
+T(\rho_{ij}\sigma_i\sigma_j+\rho_{ik}\sigma_i\sigma_k+\rho_{jk}\sigma_j\sigma_k)\right].
\end{aligned}$$

The three drift corrections again cancel the individual variance terms. Thus

$$\begin{aligned}
\mathbb E[F_i(T)F_j(T)F_k(T)]
&=F_i(0)F_j(0)F_k(0)\\
&\quad\times\exp\!\left[T(\rho_{ij}\sigma_i\sigma_j+\rho_{ik}\sigma_i\sigma_k+\rho_{jk}\sigma_j\sigma_k)\right].
\end{aligned}$$

Expanding the basket cube gives

$$\boxed{\begin{aligned}
\mathbb E[B_T^3]
&=\sum_{i=1}^n\sum_{j=1}^n\sum_{k=1}^n
w_iw_jw_kF_i(0)F_j(0)F_k(0)\\
&\quad\times\exp\!\left[T(\rho_{ij}\sigma_i\sigma_j+\rho_{ik}\sigma_i\sigma_k+\rho_{jk}\sigma_j\sigma_k)\right].
\end{aligned}}$$

All sums include repeated indices, so no additional counting factors are needed.


## 5. Moments of the approximation

The normal MGF gives the three exponential expectations directly:

$$\begin{aligned}
\mathbb E\!\left[e^{-\widehat\sigma^2T/2+\widehat\sigma\sqrt T Z}\right]&=1,\\
\mathbb E\!\left[e^{-\widehat\sigma^2T+2\widehat\sigma\sqrt T Z}\right]&=e^{\widehat\sigma^2T},\\
\mathbb E\!\left[e^{-3\widehat\sigma^2T/2+3\widehat\sigma\sqrt T Z}\right]&=e^{3\widehat\sigma^2T}.
\end{aligned}$$

Expanding the first three powers of $\widehat B_T$ therefore yields

$$\boxed{\begin{aligned}
\mathbb E[\widehat B_T]&=\widehat C+\widehat F,\\
\mathbb E[\widehat B_T^2]&=\widehat C^2+2\widehat C\widehat F
+\widehat F^2e^{\widehat\sigma^2T},\\
\mathbb E[\widehat B_T^3]&=\widehat C^3+3\widehat C^2\widehat F
+3\widehat C\widehat F^2e^{\widehat\sigma^2T}
+\widehat F^3e^{3\widehat\sigma^2T}.
\end{aligned}}$$


## 6. Equate the moments and solve

Equating the raw moments above is equivalent to equating the mean, variance, and third central moment:

$$\begin{aligned}
\mathbb E[B_T]&=\widehat C+\widehat F,\\
\operatorname{Var}(B_T)&=\widehat F^2\left(e^{\widehat\sigma^2T}-1\right),\\
\mathbb E[(B_T-\mathbb E[B_T])^3]
&=\widehat F^3\left(e^{3\widehat\sigma^2T}-3e^{\widehat\sigma^2T}+2\right).
\end{aligned}$$

From the variance equation,

$$e^{\widehat\sigma^2T}=1+\frac{\operatorname{Var}(B_T)}{\widehat F^2}.$$

Substitute this into the third-moment equation and expand:

$$\begin{aligned}
\mathbb E[(B_T-\mathbb E[B_T])^3]
&=\widehat F^3\left[\left(1+\frac{\operatorname{Var}(B_T)}{\widehat F^2}\right)^3
-3\left(1+\frac{\operatorname{Var}(B_T)}{\widehat F^2}\right)+2\right]\\
&=\frac{3\operatorname{Var}(B_T)^2}{\widehat F}
+\frac{\operatorname{Var}(B_T)^3}{\widehat F^3}.
\end{aligned}$$

For positive variance and positive third central moment, the right side decreases strictly from infinity to zero as $\widehat F$ increases. Its unique positive solution is

$$\boxed{\widehat F=
\frac{\sqrt{\operatorname{Var}(B_T)}}
{2\sinh\!\left[\frac13\operatorname{arsinh}\!\left(
\frac{\mathbb E[(B_T-\mathbb E[B_T])^3]}{2\operatorname{Var}(B_T)^{3/2}}
\right)\right]}.}$$

This is the explicit solution of the cubic above, using the triple-angle identity for $\sinh$. The other two parameters follow directly:

$$\boxed{\widehat C=\sum_{i=1}^n w_iF_i(0)-\widehat F,\qquad
\widehat\sigma=\sqrt{\frac1T\log\!\left(1+\frac{\operatorname{Var}(B_T)}{\widehat F^2}\right)}.}$$

The third central moment needed here is obtained from the three basket moments already calculated:

$$\mathbb E[(B_T-\mathbb E[B_T])^3]
=\mathbb E[B_T^3]-3\mathbb E[B_T]\mathbb E[B_T^2]+2\mathbb E[B_T]^3.$$


## 7. Conditions

The solution above assumes $\operatorname{Var}(B_T)>0$ and $\mathbb E[(B_T-\mathbb E[B_T])^3]>0$, with $\widehat F>0$ and $\widehat\sigma\geq0$.

For zero variance, use the constant fit $\widehat C=\mathbb E[B_T]$, $\widehat F=\widehat\sigma=0$. With positive variance but zero or negative third central moment, no finite positive-scale shifted lognormal matches all three moments.

Matching three moments does not guarantee matching tails or support: the fitted distribution has lower endpoint $\widehat C$. These discrepancies are tested in **Part 2.ipynb**.
